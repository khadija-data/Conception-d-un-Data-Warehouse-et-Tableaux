```python
df=pd.read_csv(r"C:\Users\user\Desktop\data\staging\darkom_annonces.csv")
```

```python
def convertir_types(df):
    df = df.copy()

    df["date_publication"] = pd.to_datetime(
        df["date_publication"],
        errors="coerce"
    )

    colonnes_numeriques = [
        "prix",
        "surface",
        "nb_chambres",
        "nb_salles_bain",
        "etage",
        "annee_construction",
    ]

    for col in colonnes_numeriques:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return df

df = convertir_types(df)
```
